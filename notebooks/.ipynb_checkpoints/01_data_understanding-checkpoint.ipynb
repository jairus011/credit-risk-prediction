# Key Findings

## Dataset Overview

The German Credit dataset contains **1,000 loan applicants** and **11 variables**, including the target variable **Risk**, which classifies applicants as either **Good** or **Bad** credit risks.

The target variable is imbalanced, with **700 Good** credit applicants and **300 Bad** credit applicants. Additionally, the dataset contains missing values in the **Saving accounts** and **Checking account** variables, which will need to be addressed during the preprocessing stage.

---

## Univariate Analysis

The exploratory analysis showed that most applicants are between **20 and 40 years** old, with the age distribution being slightly right-skewed. Both **Credit Amount** and **Loan Duration** are also right-skewed, indicating that most applicants borrow relatively smaller amounts over shorter repayment periods, while a few customers take substantially larger loans and longer repayment periods.

Most applicants:
- Own their homes.
- Belong to the **Skilled (Job Category 2)** group.
- Borrow mainly for **Car** and **Radio/TV** purchases.
- Have **Little** savings and checking account balances.

The identified outliers in Age, Credit Amount, and Duration appear to be realistic observations rather than data entry errors; therefore, they were retained for further analysis.

---

## Bivariate Analysis

Comparing each feature with the target variable revealed that **Loan Duration** appears to provide the strongest separation between Good and Bad credit applicants among the numerical variables analyzed.

**Credit Amount** also shows a moderate relationship with credit risk, with applicants classified as Bad credit generally requesting slightly larger loans than Good credit applicants.

Among the categorical variables, **Checking Account**, **Loan Purpose**, **Housing**, **Saving Accounts**, and **Job Category** provide useful information for distinguishing credit risk, although none of these variables perfectly separate the two classes.

Conversely, **Age** and **Sex** appear to have relatively weak relationships with credit risk based on the visual analysis.

---

## Correlation Analysis

The correlation analysis indicates a **moderate positive correlation (0.62)** between **Credit Amount** and **Loan Duration**, suggesting that customers requesting larger loans generally receive longer repayment periods.

The remaining numerical variables exhibit weak correlations, indicating that they capture different information. Consequently, no numerical feature appears sufficiently correlated to justify removal before model development.

---

## Conclusion

The exploratory analysis suggests that credit risk is influenced by multiple customer characteristics rather than a single variable. While Loan Duration and Credit Amount appear to be among the more informative predictors, other financial characteristics such as Checking Account, Savings Account, Housing, Job Category, and Loan Purpose also contribute useful information.

These findings provide a solid foundation for the next stage of the project, where the data will be cleaned, preprocessed, and prepared for machine learning models to predict customer credit risk.
