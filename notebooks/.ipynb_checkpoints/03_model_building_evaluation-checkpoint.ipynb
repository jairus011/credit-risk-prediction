## Load preprocessed dataset

```python
import pandas as pd
import numpy as np
# Load preprocessed features
X_train = pd.read_csv("../data/processed/X_train.csv")
X_test = pd.read_csv("../data/processed/X_test.csv")
# Load preprocessed target varriables
y_train = pd.read_csv("../data/processed/y_train.csv")

y_test = pd.read_csv("../data/processed/y_test.csv")

# convert target varriables to series
y_train = y_train.squeeze()

y_test = y_test.squeeze()

# Verify data shapes
print("Training Features:", X_train.shape)

print("Testing Features:", X_test.shape)

print("Training Target:", y_train.shape)

print("Testing Target:", y_test.shape)

# Display first five rows 
X_train.head()
```

```python
X_train.columns
```

```python
print(X_train.columns.tolist())
```

# 1.Logistic Regression (Baseline Model)

```python
# Import Logistic Regression
from sklearn.linear_model import LogisticRegression

# Create Logistic Regression model
log_reg = LogisticRegression(random_state=42)

# Train the model
log_reg.fit(X_train, y_train)
```

```python
print("Logistic Regression model trained successfully.")
```

## i.Model Prediction & Evaluation

```python
# Predict customer credit risk
y_pred = log_reg.predict(X_test)
```

```python
# Predict probability of belonging to each class
y_prob = log_reg.predict_proba(X_test)
```

```python
predictions = pd.DataFrame({
    "Actual": y_test,
    "Predicted": y_pred
})

predictions.head()
```

```python
probabilities = pd.DataFrame(
    y_prob,
    columns=["Probability_Good", "Probability_Bad"]
)

probabilities.head()
```

## ii.Model Evaluation Metrics

```python
# Import evaluation metrics
from sklearn.metrics import (
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report
)

import matplotlib.pyplot as plt
```

```python
# Confusion Matrix
cm = confusion_matrix(y_test, y_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Good", "Bad"]
)

disp.plot(cmap="Blues")

plt.title("Confusion Matrix - Logistic Regression")

plt.show()
```

```python
# Accuracy
accuracy = accuracy_score(y_test, y_pred)

print(f"Accuracy: {accuracy:.4f}")
```

```python
# Precision
precision = precision_score(y_test, y_pred)

print(f"Precision: {precision:.4f}")
```

```python
# Recall
recall = recall_score(y_test, y_pred)

print(f"Recall: {recall:.4f}")
```

```python
# F1-Score
f1 = f1_score(y_test, y_pred)

print(f"F1-Score: {f1:.4f}")
```

```python
# ROC-AUC Score
roc_auc = roc_auc_score(
    y_test,
    y_prob[:, 1]
)

print(f"ROC-AUC Score: {roc_auc:.4f}")
```

```python
# Classification Report
print(classification_report(y_test, y_pred))
```

```python
print(f"Accuracy: {accuracy:.4f}")

print(f"Precision: {precision:.4f}")

print(f"Recall: {recall:.4f}")

print(f"F1-Score: {f1:.4f}")

print(f"ROC-AUC: {roc_auc:.4f}")
```

# 2.Decision Tree Classifier

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
# Create Decision Tree model
decision_tree = DecisionTreeClassifier(
    random_state=42
)
```

```python
# Train the Decision Tree
decision_tree.fit(X_train, y_train)
```

```python
print("Decision Tree model trained successfully.")
```

## i.Decision Tree Model Prediction 

```python
# Predict customer credit risk
dt_pred = decision_tree.predict(X_test)
```

```python
# Predict class probabilities
dt_prob = decision_tree.predict_proba(X_test)
```

```python
# Compare actual vs predicted values
dt_predictions = pd.DataFrame({
    "Actual": y_test,
    "Predicted": dt_pred
})

dt_predictions.head()
```

```python
# Display prediction probabilities
dt_probabilities = pd.DataFrame(
    dt_prob,
    columns=["Probability_Good", "Probability_Bad"]
)

dt_probabilities.head()
```

## ii.Decision Tree Model Evaluation 

```python
# Confusion Matrix
dt_cm = confusion_matrix(y_test, dt_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=dt_cm,
    display_labels=["Good", "Bad"]
)

disp.plot(cmap="Greens")

plt.title("Confusion Matrix - Decision Tree")

plt.show()
```

```python
# Accuracy
dt_accuracy = accuracy_score(y_test, dt_pred)

print(f"Accuracy: {dt_accuracy:.4f}")
```

```python
# Precision
dt_precision = precision_score(y_test, dt_pred)

print(f"Precision: {dt_precision:.4f}")
```

```python
# Recall
dt_recall = recall_score(y_test, dt_pred)

print(f"Recall: {dt_recall:.4f}")
```

```python
# F1-Score
dt_f1 = f1_score(y_test, dt_pred)

print(f"F1-Score: {dt_f1:.4f}")
```

```python
# ROC-AUC Score
dt_auc = roc_auc_score(
    y_test,
    dt_prob[:, 1]
)

print(f"ROC-AUC Score: {dt_auc:.4f}")
```

```python
# Classification report
print(classification_report(y_test, dt_pred))
```

```python
print(f"Accuracy: {dt_accuracy:.4f}")

print(f"Precision: {dt_precision:.4f}")

print(f"Recall: {dt_recall:.4f}")

print(f"F1-Score: {dt_f1:.4f}")

print(f"ROC-AUC Score: {dt_auc:.4f}")
```

# 3.Random Forest Classifier

```python
# Import Random Forest
from sklearn.ensemble import RandomForestClassifier
```

```python
# Create Random Forest model
random_forest = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)
```

```python
# Train the Random Forest
random_forest.fit(X_train, y_train)
```

```python
# Confirm training
print("Random Forest model trained successfully.")
```

```python
# Predict probability of belonging to each class
rf_prob = random_forest.predict_proba(X_test)
```

```python
rf_probabilities = pd.DataFrame(
    rf_prob,
    columns=["Probability_Good", "Probability_Bad"]
)

rf_probabilities.head()
```

```python
rf_accuracy = accuracy_score(y_test, rf_pred)

print(f"Accuracy: {rf_accuracy:.4f}")
```

```python
rf_f1 = f1_score(y_test, rf_pred)

print(f"F1-Score: {rf_f1:.4f}")
```

```python
rf_auc = roc_auc_score(
    y_test,
    rf_prob[:, 1]
)

print(f"ROC-AUC Score: {rf_auc:.4f}")
```

```python
print(classification_report(
    y_test,
    rf_pred
))
```

```python
print(f"Accuracy: {rf_accuracy:.4f}")
print(f"Precision: {rf_precision:.4f}")
print(f"Recall: {rf_recall:.4f}")
print(f"F1-Score: {rf_f1:.4f}")
print(f"ROC-AUC Score: {rf_auc:.4f}")
```

## Objective

Summarize the key outcomes of the project, identify the best-performing machine learning model, discuss the limitations of the current solution, and outline recommendations for future improvements.

---

## Project Summary

This project successfully developed an end-to-end machine learning pipeline for credit risk prediction using the German Credit dataset. The workflow covered every stage of a typical machine learning project, including data understanding, exploratory data analysis, data preprocessing, feature engineering, model development, evaluation, and business interpretation.

Three classification algorithms were implemented and compared:

- Logistic Regression
- Decision Tree
- Random Forest

Each model was evaluated using Accuracy, Precision, Recall, F1-Score, ROC-AUC, and the Confusion Matrix to provide a comprehensive assessment of predictive performance.

---

## Model Comparison Summary

Among the three evaluated models, the **Random Forest Classifier** demonstrated the strongest overall performance.

| Model | Key Observation |
|-------|------------------|
| Logistic Regression | Provided a simple, interpretable baseline but struggled to identify high-risk borrowers due to low recall. |
| Decision Tree | Improved recall significantly but suffered from lower precision and signs of overfitting. |
| Random Forest | Achieved the best overall balance between Accuracy, Precision, F1-Score, and ROC-AUC while maintaining better generalization than a single Decision Tree. |

Although the Random Forest achieved the highest overall performance, its recall for the Bad class indicates that a considerable number of risky borrowers would still remain undetected. Therefore, the current implementation should be considered a strong **baseline model** rather than a production-ready credit scoring system.

---

## Business Conclusion

From a business perspective, selecting a credit risk model requires balancing profitability with financial risk.

While approving more loan applicants may increase revenue, failing to identify risky borrowers can result in substantial financial losses through loan defaults.

Based on the experimental results obtained in this project, the Random Forest model provides the most suitable balance between identifying risky applicants and minimizing unnecessary rejection of creditworthy customers. Consequently, it is recommended as the preferred baseline model for future development.

---

## Limitations

Despite encouraging results, several limitations remain:

- The dataset contains only 1,000 observations, limiting the model's ability to learn complex borrower behaviour.
- The dataset is moderately imbalanced, resulting in lower recall for high-risk applicants.
- Default model hyperparameters were used without optimization.
- Limited feature engineering was performed due to the available dataset attributes.

These limitations suggest that further improvements are necessary before deploying the model in a real-world banking environment.

---

## Future Work

Future improvements to this project may include:

- Hyperparameter tuning using GridSearchCV or RandomizedSearchCV.
- Addressing class imbalance using techniques such as SMOTE or class weighting.
- Optimizing the probability threshold to improve recall for high-risk borrowers.
- Developing additional engineered features that better capture borrower behaviour.
- Evaluating advanced ensemble methods such as XGBoost, LightGBM, or CatBoost.
- Validating the models using larger and more recent credit datasets.
